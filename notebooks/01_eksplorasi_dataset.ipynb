## Cell 1 — Memastikan lokasi dataset asli

Kode ini **hanya membaca lokasi dan header**. Tidak mengubah CSV.

```python
from pathlib import Path
import csv

CURRENT_DIR = Path.cwd().resolve()
ROOT = CURRENT_DIR if (CURRENT_DIR / "dataset" / "raw" / "books_with_categories.csv").is_file() else CURRENT_DIR.parent
DATA_PATH = ROOT / "dataset" / "raw" / "books_with_categories.csv"

if not DATA_PATH.is_file():
    raise FileNotFoundError(f"CSV asli belum ditemukan: {DATA_PATH}")

with DATA_PATH.open("r", encoding="utf-8-sig", newline="") as f:
    columns = next(csv.reader(f))

print("Project :", ROOT)
print("Dataset :", DATA_PATH)
print("Kolom   :", columns)
print("[OK] CSV asli ditemukan dan tidak diubah")
```
